```python
import ollama
```

```python
response=ollama.chat(
    model="llama3.2:1b",
    messages=[{"role":"user","content":"what is rohit sharma dob"}],
    keep_alive="1h"
)
print("ollama(local) response")
print(response["message"]["content"])
```
