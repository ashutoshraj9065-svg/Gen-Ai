```python
import ollama
import os
from dotenv import load_dotenv
from groq import Groq
```

```python
load_dotenv()
groq_key=os.getenv("Groq_key")

groq_client=Groq(api_key=groq_key)

response=groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{"role":"user","content":"who is prime minister of india"}],
    temperature=0.1)
print("Groq(cloud)response:")
print(response.choices[0].message.content)
```

```python
load_dotenv()
groq_key=os.getenv("Groq_key")

groq_client=Groq(api_key=groq_key)

sample="hello i am ashu"

response=groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[{"role":"user","content":sample}],
    temperature=0.1)
print("test",sample)
print("groq prompt token count",response.usage.prompt_tokens) 
print("Rough word count",len(sample.split()))
```
