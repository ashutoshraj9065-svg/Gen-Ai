```python

import os
from dotenv import load_dotenv
from groq import Groq
```

```python
# promot engineering me Zero-shot me classify krna padta hai example deke 
load_dotenv()
groq_key=os.getenv("Groq_key")

groq_client=Groq(api_key=groq_key)
def groq_ashu(promot,temp=0.7):
                response=groq_client.chat.completions.create(
                model="openai/gpt-oss-120b",
                messages=[{"role":"user","content":promot}],
                temperature=temp
                
                )

                return response.choices[0].message.content
groq_ashu("hello as a experiense teacher plese explain random.randint in numpy and what do you this in 4 bullet point")
```
