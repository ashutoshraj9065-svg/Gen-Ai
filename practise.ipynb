```python
from groq import Groq
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
groq_key=os.getenv("Groq_key")

groq_client=Groq(api_key=groq_key)

def groq_ashu(promot):
                response=groq_client.chat.completions.create(
                model="openai/gpt-oss-120b",
                messages=[{"role":"user","content":promot}],
                temperature=0.9
                )

                print("Groq(cloud)response:")
                print(response.choices[0].message.content)
groq_ashu("which date of birth virat kohli")
print()
groq_ashu("what is llm plese define me 3 line ")
print()
groq_ashu("what is a chat.completion.create in gen ai 4 line") 
print()
# groq_ashu("what is a groq model isme kya kya hote hai groq model plese explain in hienglish") 
groq_ashu("abhi ai me kya chal rha hai latest me kuchh new news btay 5 line me hienglish")
```
